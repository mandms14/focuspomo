```python
#pandas is a library for working with large datasets and extracting BI from it

import pandas as pd
```

```python
#read the csv file into a pandas data frame

academy = pd.read_csv('academy_awards.csv')
```

```python
#show the head of the data frame

academy.head()
```

```python
#show the tail of the data frame

academy.tail()
```

```python
#pandas series based on the Category

category = pd.Series(academy['Category'])
category
```

```python
#pandas series based on the Winner

winner = pd.Series(academy['Winner'])
winner
```

```python
#list all the Category values uniquely

category.unique()
```

```python
#count all the winners by values (how many times did someone win the award?)

winner.value_counts()
```

```python
#data frame for Best Picture

best_picture = academy[academy['Category'] == 'Best Picture']
best_picture
```

```python
#data frame for Actor in a Leading Role

leading_actor = academy[academy['Category'] == 'Actor in a Leading Role']
leading_actor
```

```python
#new data frame indexed by Year

academy_by_year = academy.set_index('Year')
academy_by_year
```

```python
#locate all the winners for 1970

academy_by_year.loc[academy_by_year.index == 1970]
```

```python
#locate all the winners for 2010

academy_by_year.loc[academy_by_year.index == 2010]
```

```python
#locate all the winners from 1970 to 2010

academy_by_year.loc[1970:2010]
```
